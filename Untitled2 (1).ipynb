```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("RemedialBigData") \
    .getOrCreate()
```

```python
data = [
    (1, "Female", 30, 5, 100.0, 0),
    (2, "Male", 45, 12, 500.0, 1),
    (3, "Female", 22, 2, 50.0, 0),
    (4, "Male", 38, 8, 300.0, 1),
    (5, "Female", 50, 20, 800.0, 0)
] * 1000

columns = [
    "user_id",
    "gender",
    "age",
    "tenure",
    "monthly_charges",
    "churn"
]

df = spark.createDataFrame(data, columns)

df.show(5)
```

```python
# Menghapus data kosong
df = df.dropna()

# Menghapus data duplikat
df = df.dropDuplicates()

print("Jumlah data:")
print(df.count())
```

```python
from pyspark.ml.feature import StringIndexer, VectorAssembler

indexer = StringIndexer(
    inputCol="gender",
    outputCol="gender_encoded"
)

df_transform = indexer.fit(df).transform(df)

assembler = VectorAssembler(
    inputCols=[
        "age",
        "tenure",
        "monthly_charges",
        "gender_encoded"
    ],
    outputCol="features"
)

final_data = assembler.transform(df_transform)

final_data.select(
    "features",
    "churn"
).show(5)
```

```python
from pyspark.ml.classification import RandomForestClassifier

train, test = final_data.randomSplit([0.8, 0.2], seed=123)

rf = RandomForestClassifier(
    featuresCol="features",
    labelCol="churn"
)

model = rf.fit(train)

prediction = model.transform(test)

prediction.select(
    "prediction",
    "churn"
).show()
```

```python
from pyspark.ml.evaluation import BinaryClassificationEvaluator

evaluator = BinaryClassificationEvaluator(
    labelCol="churn"
)

akurasi = evaluator.evaluate(prediction)

print("Nilai Evaluasi :", akurasi)
```
